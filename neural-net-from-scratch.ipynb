```python
import autograd.numpy as np
from autograd import grad
```

```python
class Layer:
    def __init__(self, dim, units):
        self.units = units
        self.dim = dim
        self.params = np.random.rand(units,dim)
    def forward(self, x):
        #print(f'input shape: {x.shape}')
        y = np.matmul(self.params, x)
        #print(f'output shape: {y.shape}')
        return y
```

```python
def MSE(y1, y2):
    return sum((y1-y2)**2)/len(y1)
```

```python
def optimise(loss_func, lr:float, epochs: int, x, y, model):
    params = model.params
    def compute_loss(params):
        model.params = params
        #print(f'x: {x}')
        y_pred = model.forward(x)
        #print(f'y: {y.shape}')
        #print(f'y_pred: {y_pred.reshape(len(y)).shape}')
        return loss_func(y, y_pred.reshape(len(y)))
    for i in range(epochs):
        #print(f'compute_loss(params): {compute_loss(params)}')
        div_loss = grad(compute_loss)
        print(f'Loss: {compute_loss(params)}')
       #print(f'model_params: {model.params}')
        params -= lr*div_loss(params)
```

```python
class Net:
    def __init__(self, *args):
        self.layers = np.array(args)
        print(f'Layers: {self.layers}')
        self.params = np.array([])
        for layer in args:
            np.append(self.params,layer.params)
        #np.array(layer.params for layer in args)
    def forward(self, x):
        for i in range(len(self.layers)):
            self.layers[i].params = self.params[i]
        state = x
        for layer in self.layers:
            state = layer.forward(state)
        return state
```

```python
np.random.seed(42)

layer1 = Layer(1,4)
layer2 = Layer(4,1)
model1 = Net(layer1, layer2)
x = np.linspace(-5, 5, num=50)
y = x + np.random.normal(loc=0.0, scale=3.0, size=50)

optimise(MSE, 0.11, 200, x.reshape(1,-1), y, model1)
```
